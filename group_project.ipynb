# 一、Data cleaning

## 1、Load data and prepare

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

path = './Datasets/Raw/'
df = pd.read_csv(path + 'survey.csv')

print(f"data shape: {df.shape}")
print("\nThe first five rows of data:")
print(df.head())
```

```python
print("\ndata info:")
print(df.info())
```

```python
print("\ndata missing value count:")

total_count = len(df)
missing_count = df.isnull().sum()
missing_ratio = (missing_count / total_count) * 100

missing_info = pd.DataFrame({
    'Total': total_count,
    'Missing': missing_count,
    'Missing_Ratio(%)': missing_ratio.round(2)
})
missing_info = missing_info[missing_info['Missing'] > 0]

print(missing_info)
```
